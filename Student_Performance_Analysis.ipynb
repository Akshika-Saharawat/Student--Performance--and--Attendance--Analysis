```python
import pandas as pd
import numpy as np

df = pd.read_csv('StudentPerformanceFactors dataset.csv')

print("Dataset loaded successfully!")
print("Dataset size:", df.shape)
```

```python
# STUDENT PERFORMANCE AND ATTENDANCE ANALYSIS

print("=" * 45)
print("1. ATTENDANCE ANALYSIS")
print("=" * 45)

print("Average Attendance:", round(df['Attendance'].mean(), 2), "%")
print("Highest Attendance:", df['Attendance'].max(), "%")
print("Lowest Attendance:", df['Attendance'].min(), "%")
print("Students above 75% attendance:", (df['Attendance'] > 75).sum())


print("\n" + "=" * 45)
print("2. STUDY HOURS ANALYSIS")
print("=" * 45)

print("Average Study Hours:", round(df['Hours_Studied'].mean(), 2))
print("Maximum Study Hours:", df['Hours_Studied'].max())
print("Minimum Study Hours:", df['Hours_Studied'].min())


print("\n" + "=" * 45)
print("3. EXAM PERFORMANCE ANALYSIS")
print("=" * 45)

print("Average Exam Score:", round(df['Exam_Score'].mean(), 2))
print("Highest Exam Score:", df['Exam_Score'].max())
print("Lowest Exam Score:", df['Exam_Score'].min())
print("Students scoring above 80:", (df['Exam_Score'] > 80).sum())


print("\n" + "=" * 45)
print("4. CORRELATION ANALYSIS")
print("=" * 45)

selected_columns = ['Hours_Studied', 'Attendance',
                    'Previous_Scores', 'Exam_Score']

correlation = df[selected_columns].corr()

print(correlation['Exam_Score'].sort_values(ascending=False))


print("\n" + "=" * 45)
print("5. DATASET SUMMARY")
print("=" * 45)

print("Total Student Records:", df.shape[0])
print("Total Columns:", df.shape[1])
print("Missing Values:", df.isnull().sum().sum())
print("Duplicate Records:", df.duplicated().sum())

print("\nANALYSIS COMPLETED SUCCESSFULLY!")
```
