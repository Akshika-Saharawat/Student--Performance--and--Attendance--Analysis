# STUDENT PERFORMANCE AND ANALYSIS
This project, titled "Student Performance and Attendance Analysis Using Python", focuses on analysing student academic performance and attendance patterns using Python.

The project uses a student performance dataset containing 199 records and 20 columns. With the help of Python libraries such as Pandas and NumPy, the dataset is explored, cleaned, and analysed to extract meaningful information.

The analysis includes attendance patterns, study hours, examination scores, and correlations between selected numerical factors. Missing values are handled, duplicate records are checked, and statistical methods are used to understand the dataset.

The main aim of this project is to demonstrate how Python can be used in Data Analytics to process data, identify patterns, and draw meaningful conclusions.

Tools and Technologies: Python, Pandas, NumPy, Google Colab, and Kaggle Dataset.

```python
import pandas as pd
import numpy as np
# Load the dataset
df = pd.read_csv("StudentPerformanceFactors dataset.csv")
print("Dataset loaded successfully!")
print("Number of rows:", df.shape[0])
print("Number of columns:", df.shape[1])
# Explore the dataset
print("\nFirst five records:")
print(df.head())
print("\nDataset information:")
df.info()
print("\nStatistical summary:")
print(df.describe())
print("\nColumn names:")
print(df.columns.tolist())
# Check missing values and duplicate records
print("\nMissing values before cleaning:")
print(df.isnull().sum())
print("\nNumber of duplicate records:")
print(df.duplicated().sum())
# Handle missing values
if df["Teacher_Quality"].isnull().any():
    df["Teacher_Quality"] = df["Teacher_Quality"].fillna(
        df["Teacher_Quality"].mode()[0]
    )
if df["Distance_from_Home"].isnull().any():
    df["Distance_from_Home"] = df["Distance_from_Home"].fillna(
        df["Distance_from_Home"].mode()[0]
    )
print("\nMissing values after cleaning:")
print(df.isnull().sum())
# Attendance analysis
print("\nATTENDANCE ANALYSIS")
print("Average attendance:", round(df["Attendance"].mean(), 2))
print("Highest attendance:", df["Attendance"].max())
print("Lowest attendance:", df["Attendance"].min())
print(
    "Students with attendance above 75%:",
    (df["Attendance"] > 75).sum()
)
# Study-hours analysis
print("\nSTUDY HOURS ANALYSIS")
print("Average study hours:", round(df["Hours_Studied"].mean(), 2))
print("Maximum study hours:", df["Hours_Studied"].max())
print("Minimum study hours:", df["Hours_Studied"].min())
# Exam-performance analysis
print("\nEXAM PERFORMANCE ANALYSIS")
print("Average exam score:", round(df["Exam_Score"].mean(), 2))
print("Highest exam score:", df["Exam_Score"].max())
print("Lowest exam score:", df["Exam_Score"].min())
print("Students scoring above 80:", (df["Exam_Score"] > 80).sum())
# Correlation analysis (numerical results only; no charts)print("\nCORRELATION WITH EXAM SCORE")
correlation = df.corr(numeric_only=True)
print(
    correlation["Exam_Score"]
    .drop("Exam_Score")
    .sort_values(ascending=False)
)
# Display the final cleaned dataset size
print("\nFINAL DATASET SIZE")
print("Rows:", df.shape[0])
print("Columns:", df.shape[1])
```

The dataset contained 199 student records and 20 columns. Missing values were handled using the mode method, and no duplicate records were found.
